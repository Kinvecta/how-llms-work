## Setup: the toy vectors
Axis 1 = concrete thing, axis 2 = living, axis 3 = feeling or state. These are chosen by hand.

```python
import numpy as np

cat   = np.array([1, 1, 0], dtype=float)
mat   = np.array([1, 0, 0], dtype=float)
tired = np.array([0, 0, 1], dtype=float)

def dot(a, b):
    return float(np.sum(a * b))           # a·b = sum over i of a_i * b_i

def norm(a):
    return float(np.sqrt(dot(a, a)))      # ||a|| = sqrt(a·a)

def cosine(a, b):
    return dot(a, b) / (norm(a) * norm(b))  # cos φ = a·b / (||a|| ||b||)

def angle_degrees(a, b):
    return float(np.degrees(np.arccos(cosine(a, b))))
```

## Step 1: dot products

```python
print("cat · mat   =", dot(cat, mat))
print("cat · tired =", dot(cat, tired))
assert dot(cat, mat) == 1
assert dot(cat, tired) == 0
```

## Step 2: norms

```python
print("||cat||   =", round(norm(cat), 3), "(√2 ≈ 1.41)")
print("||mat||   =", norm(mat))
print("||tired|| =", norm(tired))
assert np.isclose(norm(cat), np.sqrt(2))
assert norm(mat) == 1 and norm(tired) == 1
```

## Step 3: cosines and angles

```python
print("cat and mat:   cos φ =", round(cosine(cat, mat), 2), "  φ =", round(angle_degrees(cat, mat), 1), "degrees")
print("cat and tired: cos φ =", round(cosine(cat, tired), 2), "  φ =", round(angle_degrees(cat, tired), 1), "degrees")
assert round(cosine(cat, mat), 2) == 0.71
assert np.isclose(angle_degrees(cat, mat), 45.0)
assert cosine(cat, tired) == 0
assert np.isclose(angle_degrees(cat, tired), 90.0)
```

## Step 4: the scaling check
cat′ points the same way as cat but is three times longer. The dot product changes. The cosine does not.

```python
cat_prime = np.array([3, 3, 0], dtype=float)

print("cat  · mat =", dot(cat, mat),        "  cosine =", round(cosine(cat, mat), 2))
print("cat′ · mat =", dot(cat_prime, mat),  "  cosine =", round(cosine(cat_prime, mat), 2))
assert dot(cat_prime, mat) == 3
assert np.isclose(cosine(cat_prime, mat), cosine(cat, mat))
```

## Check question
Using the toy vectors, which pair has the larger cosine, cat–mat or cat–tired? Replacing cat with cat′, which quantity changes: the dot product with mat, or the cosine?

```python
print("Larger cosine: cat–mat (%.2f) against cat–tired (%.2f)" % (cosine(cat, mat), cosine(cat, tired)))
print("Dot product with mat goes from %g to %g. The cosine stays at %.2f." % (dot(cat, mat), dot(cat_prime, mat), cosine(cat_prime, mat)))
```
